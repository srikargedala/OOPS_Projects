```python
from abc import ABC,abstractmethod
class Product(ABC):
    def __init__(self,name,price,stock):
        self.name=name
        self.price=price
        self.stock=stock
    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self,value):
        if value<=0:
            raise ValueError('Price must be positive')
        self.__price=value
    @property
    def stock(self):
        return self.__stock
    @stock.setter
    def stock(self,value):
        if value<1:
            raise ValueError('Enter valid stock quantity')
        self.__stock=value
    @abstractmethod
    def get_category(self):
        pass
class ElectronicsProduct(Product):
    def __init__(self,name,price,stock,warranty_years):
        super().__init__(name,price,stock)
        self.warranty_years=warranty_years
    def get_category(self):
        return 'Electronics'
class ClothingProduct(Product):
    def get_category(self):
        return 'Clothing'
class BookProduct(Product):
    def get_category(self):
        return 'Books'
class OrderItem:
    def __init__(self,product,quantity):
        self.product=product
        self.quantity=quantity
    def get_total(self):
        return self.product.price*self.quantity
class Order:
    orders=0
    def __init__(self):
        Order.orders+=1
        self.order_id=Order.orders
        self.items=[]
    def add_item(self,item):
        if item in self.items:
            raise ValueError('Item already present')
        self.items.append(item)
    def get_total(self):
        total=0
        for item in self.items:
            total+=item.get_total()
        return total
class Customer:
    def __init__(self,name,email):
        self.name=name
        self.email=email
        self.orders=[]
    def add_order(self,order):
        self.orders.append(order)
    def get_total_spent(self):
        total=0
        for order in self.orders:
            total+=order.get_total()
        return total
laptop = ElectronicsProduct("Laptop", 60000, 5, 2)
jeans = ClothingProduct("Jeans", 1800, 120)
hp = BookProduct("Harry Potter", 250, 700)

print(laptop.name)
print(laptop.price)
print(laptop.stock)
print(laptop.warranty_years)

print(jeans.name)
print(jeans.price)
print(jeans.stock)

print(hp.name)
print(hp.price)
print(hp.stock)

print(laptop.get_category())
print(jeans.get_category())
print(hp.get_category())

try:
    laptop.price = -500
except ValueError as e:
    print(e)

try:
    laptop.stock = 0
except ValueError as e:
    print(e)

item1 = OrderItem(laptop, 2)
item2 = OrderItem(jeans, 3)

print(item1.product.name)
print(item1.quantity)
print(item1.get_total())

print(item2.product.name)
print(item2.quantity)
print(item2.get_total())

order1 = Order()

order1.add_item(item1)
order1.add_item(item2)

print(order1.order_id)
print(order1.get_total())

try:
    order1.add_item(item1)
except ValueError as e:
    print(e)

customer = Customer("Srikar", "srikar@example.com")

customer.add_order(order1)

print(customer.name)
print(customer.email)
print(customer.get_total_spent())

order2 = Order()

item3 = OrderItem(hp, 4)

order2.add_item(item3)

customer.add_order(order2)

print(order1.get_total())
print(order2.get_total())
print(customer.get_total_spent())

print(order1.order_id)
print(order2.order_id)

order3 = Order()
print(order3.order_id)
```
