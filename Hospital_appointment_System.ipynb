```python
class InvalidTimeError(Exception):
    def __init__(self,time):
        self.time=time
        super().__init__(f'{time} is not available')
class Person:
    def __init__(self,name,age,gender,blood_group):
        self.name=name
        self.age=age
        self.gender=gender
        self.blood_group=blood_group
class Patient(Person):
    def __init__(self,pt_id,medical_issue,ph_num,name,age,gender,blood_group):
        self.pt_id=pt_id
        self.medical_issue=medical_issue
        self.ph_num=ph_num
        super().__init__(name,age,gender,blood_group)
class Doctor(Person):
    def __init__(self,dt_id,specialization,experience,consultation_fee,name,gender,age,blood_group):
        self.available_slots=["10:00 AM", "11:00 AM", "2:00 PM", "4:00 PM"]
        self.dt_id=dt_id
        self.specialization=specialization
        self.experience=experience
        self.consultation_fee=consultation_fee
        super().__init__(name,age,gender,blood_group)
    def book_slot(self,time):
        if time not in self.available_slots:
            raise InvalidTimeError(time)
        print('Slot booked successfully')
        self.available_slots.remove(time)
class Appointment:
    def __init__(self,apt_id,doctor,patient,date,time):
        self.apt_id=apt_id
        self.doctor=doctor
        self.patient=patient
        self.date=date
        self.time=time
    def display_details(self):
        print('Appointement Details:')
        print('----------------------')
        print(f'Appointment ID: {self.apt_id}')
        print(f"Patient's name: {self.patient.name}")
        print(f"Doctor's name: {self.doctor.name}")
        print(f"Doctor's specialization: {self.doctor.specialization}")
        print(f'Date: {self.date}')
        print(f'Time: {self.time}')
class Hospital:
    def __init__(self,name):
        self.appointment_count=0
        self.name=name
        self.doctors=[]
        self.patients=[]
        self.appointments=[]
    def add_doctor(self,doctor):
        if doctor in self.doctors:
            raise ValueError('Doctor already present')
        self.doctors.append(doctor)
    def add_patient(self,patient):
        if patient in self.patients:
            raise ValueError('Patient is already registered')
        self.patients.append(patient)
    def book_appointment(self,patient,doctor,date,time):
        if patient not in self.patients:
            raise ValueError('Patient not registered')
        if doctor not in self.doctors:
            raise ValueError('Doctor not registered')
        doctor.book_slot(time)
        self.appointment_count+=1
        appointment=Appointment(self.appointment_count,doctor,patient,date,time)
        self.appointments.append(appointment)
        return appointment
    def cancel_appointment(self,appointment):
        if appointment not in self.appointments:
            raise ValueError('Appointement is invalid')
        doctor=appointment.doctor
        time=appointment.time
        doctor.available_slots.append(time)
        self.appointments.remove(appointment)
        print('Appointment cancelled successfully')
# -----------------------------
# CREATE OBJECTS
# -----------------------------

hospital = Hospital("Apollo Hospital")

patient1 = Patient(
    101,
    "Fever",
    "9876543210",
    "Rahul",
    25,
    "Male",
    "O+"
)

doctor1 = Doctor(
    201,
    "General Physician",
    8,
    800,
    "Dr. Sharma",
    42,
    "Male",
    "B+"
)


# -----------------------------
# REGISTER PATIENT & DOCTOR
# -----------------------------

hospital.add_patient(patient1)
hospital.add_doctor(doctor1)

print("Patient and Doctor registered successfully")


# -----------------------------
# CHECK AVAILABLE SLOTS
# -----------------------------

print("\nAvailable slots before booking:")
print(doctor1.available_slots)


# -----------------------------
# BOOK APPOINTMENT
# -----------------------------

appointment1 = hospital.book_appointment(
    patient1,
    doctor1,
    "20-08-2026",
    "10:00 AM"
)


# -----------------------------
# DISPLAY APPOINTMENT
# -----------------------------

print()
appointment1.display_details()


# -----------------------------
# CHECK SLOTS AFTER BOOKING
# -----------------------------

print("\nAvailable slots after booking:")
print(doctor1.available_slots)


# -----------------------------
# CANCEL APPOINTMENT
# -----------------------------

hospital.cancel_appointment(appointment1)

print("\nAppointment cancelled successfully")


# -----------------------------
# CHECK SLOTS AFTER CANCELLATION
# -----------------------------

print("\nAvailable slots after cancellation:")
print(doctor1.available_slots)


# -----------------------------
# CHECK APPOINTMENTS
# -----------------------------

print("\nHospital appointments:")
print(hospital.appointments)
```
