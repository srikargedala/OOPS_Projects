# Library Management System

```python
class Book:
    def __init__(self,book_id,title,author,category):
        self.book_id=book_id
        self.title=title
        self.author=author
        self.category=category
        self.available=True
class Member:
    def __init__(self,member_id,name):
        self.member_id=member_id
        self.name=name
        self.borrowed_books=[]
    def borrow_book(self,book):
        if not book.available:
            raise ValueError('Book is not available')
        self.borrowed_books.append(book)
        book.available=False
    def return_book(self,book):
        if book not in self.borrowed_books:
            raise ValueError('Book was not borrowed')
        self.borrowed_books.remove(book)
        self.available=True
class Library:
    def __init__(self,name):
        self.name=name
        self.catalog=LibraryCatalog()
        self.books=[]
    def add_book(self,book):
        self.books.append(book)
        self.catalog.add_book(book)
    def remove_book(self,book_id):
        for book in self.books:
            if book.book_id==book_id:
                self.books.remove(book)
    def find_book(self,book_id):
        for book in self.books:
            if book.book_id==book_id:
                return book
        return None
class LibraryCatalog:
    def __init__(self):
        self.categories={}
    def add_book(self,book):
        if book.category not in self.categories:
            self.categories[book.category]=[]
        self.categories[book.category].append(book)
class Librarian:
    def __init__(self,librarian_id,name):
        self.librarian_id=librarian_id
        self.name=name
    def issue_book(self,member,book):
        member.borrow_book(book)
    def receive_book(self,member,book):
        member.return_book(book)
book1 = Book(101, "Python Basics", "John", "Programming")
book2 = Book(102, "Data Structures", "Mark", "Programming")
book3 = Book(103, "Physics", "David", "Science")

library = Library('Central Library')

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

member = Member(1, "Srikar")

librarian = Librarian(1, "Robert")

librarian.issue_book(member, book1)

print(book1.available)
print(member.borrowed_books[0].title)

librarian.receive_book(member, book1)

print(book1.available)
```
